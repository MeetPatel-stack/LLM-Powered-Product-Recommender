```python
import json
from tqdm.notebook import tqdm

fashion_file = "../data2/meta_fashion.jsonl"
electronics_file = "../data2/meta_Electronics.jsonl"

output_file = "../data2/processed/meta_cleaned_combined.jsonl"

remove_fields = {
    "images",
    "videos",
    "store",
    "categories",
    "bought_together"
}


def count_lines(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        return sum(1 for line in f if line.strip())


fashion_total = count_lines(fashion_file)
electronics_total = count_lines(electronics_file)

total_products = fashion_total + electronics_total

print(f"Fashion products     : {fashion_total:,}")
print(f"Electronics products : {electronics_total:,}")
print(f"Total products       : {total_products:,}")


with open(output_file, "w", encoding="utf-8") as outfile:

    # Fashion
    with open(fashion_file, "r", encoding="utf-8") as infile:
        for line in tqdm(
            infile,
            total=fashion_total,
            desc="Processing Fashion",
            colour="green"
        ):
            if not line.strip():
                continue

            product = json.loads(line)

            for field in remove_fields:
                product.pop(field, None)

            outfile.write(
                json.dumps(product, ensure_ascii=False) + "\n"
            )

    # Electronics
    with open(electronics_file, "r", encoding="utf-8") as infile:
        for line in tqdm(
            infile,
            total=electronics_total,
            desc="Processing Electronics",
            colour="green"
        ):
            if not line.strip():
                continue

            product = json.loads(line)

            for field in remove_fields:
                product.pop(field, None)

            outfile.write(
                json.dumps(product, ensure_ascii=False) + "\n"
            )

print("\n✅ Combined metadata created!")
print(f"Output file: {output_file}")
```

```python

input_file = "../data2/processed/meta_cleaned_combined.jsonl"
output_file = "../data2/processed/meta_filtered.jsonl"

# Count total products so tqdm can show proper progress
with open(input_file, "r", encoding="utf-8") as f:
    total_products = sum(1 for line in f if line.strip())

# Counters
removed_zero_reviews = 0
removed_null_price = 0
removed_missing_asin = 0
total_removed = 0

with open(input_file, "r", encoding="utf-8") as infile, \
     open(output_file, "w", encoding="utf-8") as outfile:

    for line in tqdm(
        infile,
        total=total_products,
        desc="Filtering products",
        colour="green"
    ):
        if not line.strip():
            continue

        product = json.loads(line)

        parent_asin = product.get("parent_asin")
        price = product.get("price")
        review_count = product.get("rating_number")

        # Check every removal condition
        missing_asin = (
            parent_asin is None or
            str(parent_asin).strip() == ""
        )

        null_price = (
            price is None or
            str(price).strip() == ""
        )

        zero_reviews = (
            review_count is not None and
            review_count <= 0
        )

        # Count each reason independently
        if zero_reviews:
            removed_zero_reviews += 1

        if null_price:
            removed_null_price += 1

        if missing_asin:
            removed_missing_asin += 1

        # Remove product if any condition is true
        if zero_reviews  or null_price or  missing_asin:
            total_removed += 1
            continue

        # Keep valid product
        outfile.write(
            json.dumps(product, ensure_ascii=False) + "\n"
        )

print("\nFiltering completed!")
print(f"Products removed because of 0 reviews : {removed_zero_reviews}")
print(f"Products removed because of null price : {removed_null_price}")
print(f"Products removed because of missing ASIN: {removed_missing_asin}")
print(f"Total unique products removed         : {total_removed}")
print(f"Products remaining                    : {total_products - total_removed}")
```

```python
meta_file = "../data2/processed/meta_filtered.jsonl"
reviews_file = "../data2/processed/reviews_cleaned_combined.jsonl"
output_file = "../data2/processed/reviews_filtered.jsonl"

valid_asins = set()

with open(meta_file, "r", encoding="utf-8") as f:
    for line in tqdm(f, desc="Loading valid product ASINs", colour="green"):
        if not line.strip():
            continue

        product = json.loads(line)

        parent_asin = product.get("parent_asin")

        if parent_asin:
            valid_asins.add(parent_asin)

print(f"Valid product ASINs: {len(valid_asins):,}")
```

```python
# First count total reviews for a proper progress bar
with open(reviews_file, "r", encoding="utf-8") as f:
    total_reviews = sum(1 for line in f if line.strip())

kept_reviews = 0
removed_reviews = 0
invalid_json = 0

with open(reviews_file, "r", encoding="utf-8") as infile, \
     open(output_file, "w", encoding="utf-8") as outfile:

    for line in tqdm(
        infile,
        total=total_reviews,
        desc="Filtering reviews",
        colour="green"
    ):
        if not line.strip():
            continue

        try:
            review = json.loads(line)
        except json.JSONDecodeError:
            invalid_json += 1
            continue

        parent_asin = review.get("parent_asin")

        if parent_asin in valid_asins:
            outfile.write(
                json.dumps(review, ensure_ascii=False) + "\n"
            )
            kept_reviews += 1
        else:
            removed_reviews += 1

print("\nFiltering completed!")
print(f"Total reviews    : {total_reviews:,}")
print(f"Reviews kept     : {kept_reviews:,}")
print(f"Reviews removed  : {removed_reviews:,}")
print(f"Invalid JSON     : {invalid_json:,}")
```

```python
from collections import Counter
reviews_file = "../data2/processed/reviews_filtered.jsonl"

user_review_count = Counter()

# Count total reviews for progress bar
with open(reviews_file, "r", encoding="utf-8") as f:
    total_reviews = sum(1 for line in f if line.strip())

# Count reviews per user
with open(reviews_file, "r", encoding="utf-8") as f:
    for line in tqdm(
        f,
        total=total_reviews,
        desc="Counting user reviews",
        colour="green"
    ):
        if not line.strip():
            continue

        review = json.loads(line)

        user_id = review.get("user_id")

        if user_id:
            user_review_count[user_id] += 1

# Users having more than 2 reviews
users_more_than_3 = sum(
    1 for count in user_review_count.values()
    if count > 3
)

print(f"Total unique users       : {len(user_review_count):,}")
print(f"Users with > 3 reviews   : {users_more_than_3:,}")
```

```python
import json
from collections import defaultdict
from tqdm.notebook import tqdm

reviews_file = "../data2/processed/reviews_filtered.jsonl"

user_products = defaultdict(set)

# Count total reviews
with open(reviews_file, "r", encoding="utf-8") as f:
    total_reviews = sum(1 for line in f if line.strip())

# Collect distinct products for each user
with open(reviews_file, "r", encoding="utf-8") as f:
    for line in tqdm(
        f,
        total=total_reviews,
        desc="Finding distinct products per user",
        colour="green"
    ):
        if not line.strip():
            continue

        review = json.loads(line)

        user_id = review.get("user_id")
        parent_asin = review.get("parent_asin")

        if user_id and parent_asin:
            user_products[user_id].add(parent_asin)

print("\nUsers by number of DISTINCT products:\n")

for threshold in [2, 3, 4, 5, 10, 20]:
    count = sum(
        1 for products in user_products.values()
        if len(products) >= threshold
    )

    print(f">= {threshold} distinct products : {count:,}")
```

```python
eligible_users = [
    user_id
    for user_id, products in user_products.items()
    if len(products) >= 6
]

print("Eligible users:", len(eligible_users))
```

```python
import random
random.seed(42)

selected_users = set(
    random.sample(eligible_users, 200_000)
)

print("Selected users:", len(selected_users))
```

```python
user_interactions = defaultdict(list)

with open(reviews_file, "r", encoding="utf-8") as f:
    for line in tqdm(f, desc="Collecting interactions", colour="green"):
        review = json.loads(line)

        user_id = review.get("user_id")

        if user_id in selected_users:
            user_interactions[user_id].append({
                "parent_asin": review.get("parent_asin"),
                "rating": review.get("rating"),
                "timestamp": review.get("timestamp")
            })
```

```python
import json

output_interactions_file = "../data2/processed/final_user_interactions.jsonl"

# Save the defaultdict to JSONL (One user per line)
with open(output_interactions_file, "w", encoding="utf-8") as f:
    for user_id, interactions in user_interactions.items():
        row = {
            "user_id": user_id,
            "interactions": interactions
        }
        f.write(json.dumps(row) + "\n")
```
